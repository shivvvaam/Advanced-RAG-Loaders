```python
from langchain_community.document_loaders import WebBaseLoader
from pprint import pp
```

```python
#create a loader
url_1="https://reference.langchain.com/python/langchain-community/document_loaders/pdf/PyPDFLoader"
url_2="https://reference.langchain.com/python/langchain-community/document_loaders/text/TextLoader"
url_3="https://reference.langchain.com/python/langchain-community/document_loaders/json_loader/JSONLoader"
loader = WebBaseLoader(web_paths=[url_1,url_2,url_3])
```

```python
#load the documnets
documents = loader.load()
```

```python
print(documents[2].page_content)
```

```python
pp(documents[1].metadata)
```

## RecursiveUrlLoader ##

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
#create a  loader
base_url = "https://reference.langchain.com/python/langchain-community/document_loaders"
loader = RecursiveUrlLoader(url=base_url,  max_depth=2)
```

```python
#load the documents
documents = loader.load()
```

```python
len(documents)
```

```python
#print the metadata for first 10 documents
for i in range(10):
    print(f"Document {i+1} metadata:")
    pp(documents[i].metadata)
    print("\n")
```

## lazy load ##
### we don't hav eto wait for complete executio .It displays as its gets executed ###

```python
#lazy load
documents_lazy_load = loader.lazy_load()
```

```python
print(documents[0].page_content[:500])
```

```python
documents_lazy_load
```

```python
counter=0
for doc in documents_lazy_load:
    if counter==20:
        break
    print(f"Document {counter+1} page content:")
    print(doc.page_content[:300])
    pp(doc.metadata)
    print("\n")
    counter+=1
```
